# Recursion

```python
def printnum(n):
    if n>0:
        printnum(n-1)
        print(n)
```

```python
printnum(20)
```

```python
def sumnumbers(n):
    total =0
    for i in range(1,n+1):
        total =total+i
    return total
```

```python
sumnumbers(5)
```

```python
def sumNumberRecrsivly(n):
    if n == 1 : return 1
    return n + sumNumberRecrsivly(n-1)

```

```python
sumNumberRecrsivly(5)
```
